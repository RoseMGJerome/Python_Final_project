# **Data cleaning**

### **Identifying duplicates data**

```python
# Identifying duplicates
df2.duplicated().value_counts()
```

The code above shows that 81 records are duplicated out of the 35,400 entries. Let’s take a closer look by selecting specific columns to determine whether additional duplicate records exist.

```python
#Checking of duplicates by subset

subset = df2[["NSM", "Usage_kWh", 'Day_of_week']]

duplicates = subset[subset.duplicated(keep=False)]

print(duplicates)
```

In the code above and below, we used three columns (NSM, Usage_kWh, and Day_of_week) to identify duplicate records in the dataset. The results reveal a considerable number of duplicate records based on these three columns.

```python
df2.loc[ df2.duplicated(subset=['NSM', 'Usage_kWh', 'Day_of_week'])]
```

### **Dropping duplicates**

```python
df2.drop_duplicates(subset=['NSM', 'Usage_kWh', 'Day_of_week'], inplace=True)
df2.head()
```

## **Dropping null values**

```python
df2.dropna(inplace=True)
```

# **Identifying outliers**

```python
#Identifying outliers
# Mean and Standard deviation of Usag_KWH
average_usage=df2['Usage_kWh'].mean()
std_usage=df2['Usage_kWh'].std()
print(average_usage)
print(std_usage)
```

```python
# Calculation of the Z-score
df2['z_score']=(df2['Usage_kWh']-average_usage)/std_usage
df2
```

```python
df2[df2['z_score']>3]
```

Comparing the Z-scores against the threshold of 3 reveals that the dataset contains potential outliers (Z-score > 3).

Identifying and handling outliers is an important step in data analytics because they can distort statistical analyses, reduce model accuracy, and lead to misleading conclusions.

# **Data manipulation & Visualization**

```python
#create a new column "year" from date, coercing invalid formats like '08/1' to NaT
df2['date'] = pd.to_datetime(df2['date'], format='mixed', dayfirst=True, errors='coerce')
df2['year'] = df2['date'].dt.year
df2
```

```python
#create a new column "Month" from date

df2['month'] = df2['date'].dt.month
df2
```

## **How does electricity consumption differ between weekdays and weekends?**

```python
# Average electricity consumption on weekdays
Average_electricity_consumption_on_weekdays=df2[df2['Day_of_week'].isin(["Monday","Tuesday","Wednesday","Thursday","Friday"])]['Usage_kWh'].mean()
Average_electricity_consumption_on_weekdays
```

```python
# Average electricity consumption on weekends
Average_electricity_consumption_on_weekends=df2[df2['Day_of_week'].isin(["Saturday","Sunday"])]['Usage_kWh'].mean()
Average_electricity_consumption_on_weekends
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(data=df2, x='Day_of_week', y='Usage_kWh', hue="Day_of_week")
plt.xticks(rotation=90)
plt.xlabel('Day of the Week')
plt.ylabel('Usage (kWh)')
plt.title('Electricity Consumption by Day of the Week')
plt.show()
```

# **Which month has the highest average consumption?**

```python
df2.groupby('month')['Usage_kWh'].mean().sort_values(ascending=False)
```

```python
# Create a bar plot month to represente the variation of the average consumption over the month
sns.barplot(data=df2, x='month', y='Usage_kWh',errorbar=None,)
plt.title('Average Electricity Consumption by Month')
plt.xlabel('Month')
plt.ylabel('Average Consumption (kWh)')
plt.show()
```

January has the highest average energy consumption (kWh), followed by February. This may indicate that the company received more orders at the beginning of the year, possibly due to increased demand during the holiday season compared to other months.

## **Why is Load_Type important for our analysis?**
Load_Type helps you understand how the factory's electricity demand changes according to its operating conditions.

This visualization shows that the Light_Load has a higher energy demand than the other load types.

## Understanding the three Load_Type categories
### Light_Load (Low demand)

The factory requires a relatively small amount of electricity. This could happen when fewer machines are operating, production activity is lower, or some equipment is not being used.

### Medium_Load (Moderate demand)

The factory requires a moderate amount of electricity. More machines and equipment may be operating, and production activity may be at a normal or intermediate level.

### Maximum_Load (High demand)

The factory requires a large amount of electricity. This may happen when several energy-intensive machines or production processes are operating simultaneously.

### **line plot to visualize energy consumption by hour and days of the week. This can help identify when the factory consumes the most electricity.**

### **When does the factory consume the most electricity?**

```python
#Average energy consume by hour
import matplotlib.pyplot as plt
import seaborn as sns

df2['Hour']=df2['NSM']//3600
# Group by both Hour and Day_of_week to preserve the Day_of_week column for the plot hue
Average_electricity_hour=df2.groupby(['Hour', 'Day_of_week'])['Usage_kWh'].mean().reset_index()

plt.figure(figsize=(12, 5))
#plt.style.use('fivethirtyeight')
sns.lineplot(data=Average_electricity_hour, x='Hour', y='Usage_kWh', hue='Day_of_week', marker="o")
plt.title('Hourly Energy Consumption by Day of the Week')
plt.xlabel('Hour')
plt.ylabel('Average Usage (kWh)')
plt.xticks(range(24))
plt.show()
```

## **Insights**
For this visualization, we chose a line plot to identify when the factory consumes the most electricity during the day and for differents days of the week.
According to the graph, we can see that electricity consumption is high between 7:00 AM and 12:00 PM, decreases between 12:00 PM and 1:00 PM, and then increases significantly from 1:00 PM to 8:00 PM.
This pattern might suggest that the factory operates from 7:00 AM to 8:00 PM, with a long break around noon.

This is a useful visualization because it helps us understand how average electricity consumption is distributed throughout the day.


We observe lower energy consumption during the weekend, especially on Sundays. This may indicate that the company is closed on weekends or operates only a limited number of machines during those days.
